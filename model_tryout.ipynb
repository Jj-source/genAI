# Model Tests

```python
import requests
import time

url = "http://localhost:11434/api/generate"

with open('system_prompt_republican.txt', 'r') as file:
    system_prompt_rep = file.read()

with open('system_prompt_democratic.txt', 'r') as file:
    system_prompt_dem = file.read()

with open('constitution_republican.txt', 'r') as file:
    const_rep = file.read()

with open('constitution_democratic.txt', 'r') as file:
    const_dem = file.read()

with open('system_instruction_rep.txt', 'r') as file:
    system_instruction_rep = file.read()

with open('system_instruction_dem.txt', 'r') as file:
    system_instruction_dem = file.read()
```

## Republican

### With constitution

```python
user_promt = "Should women have the right for abortion?"

prompt_text = f"System: {system_prompt_rep} {const_rep}\n User: {user_promt}"

payload = {
    "model": "ikiru/dolphin-mistral-24b-venice-edition",
    "prompt": prompt_text,
    "stream": False
}

# Retry in case model is still loading
for _ in range(3):
    response = requests.post(url, json=payload)
    response.raise_for_status()
    data = response.json()
    if data["response"]:
        print(data["response"])
        break
    else:
        print("Model still loading, retrying...")
        time.sleep(2)
```
